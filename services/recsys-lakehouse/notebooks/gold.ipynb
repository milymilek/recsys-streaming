```python
from recsys_lakehouse.spark import spark_builder
from recsys_lakehouse.lakehouse.operator import TableOperator
from recsys_lakehouse.lakehouse import layers

from pathlib import Path
from pyspark.sql import functions as F
from pyspark.sql.functions import from_unixtime, col, sum, rand, when
from pyspark.sql.types import FloatType, IntegerType, TimestampType
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("").getOrCreate()  # type: ignore
```

```python
bronze_layer = layers.Bronze(dataset_name="amazon_books_sample10000")
silver_layer = layers.Silver(dataset_name="amazon_books_sample10000")
operator = TableOperator(spark)
```

```python
rev = operator.read_table(silver_layer, silver_layer.tables['books_reviews'])
```

```python
meta = operator.read_table(silver_layer, silver_layer.tables['books_metadata'])
```

```python
from pyspark.sql import Window

df = rev.select("user_id", "title", "text", "timestamp")
windowSpec = Window.partitionBy("user_id").orderBy("timestamp")

rev_with_row_num = df.withColumn("row_num", F.row_number().over(windowSpec))

rev_aggregated = rev_with_row_num.groupBy("user_id").agg(
    F.collect_list(F.struct("title", "text", "timestamp")).alias("reviews")
)

rev_aggregated.show(truncate=False)
```

```python
rev_filtered = rev.filter(col("timestamp") < "2023-01-01")
rev_grouped = rev_filtered.groupBy("user_id").agg(
    F.count("*").alias("n_reviews"),
    F.mean("rating").alias("mean_rating"),
    F.mean("helpful_vote").alias("mean_helpful_vote"),
    F.stddev("helpful_vote").alias("std_helpful_vote")
).fillna({'std_helpful_vote': 0})

rev_grouped.show()
```
