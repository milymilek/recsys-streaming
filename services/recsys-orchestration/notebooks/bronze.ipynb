```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, from_unixtime, date_format
```

```python
spark = SparkSession.builder \
    .master("spark://localhost:7077") \
    .appName("PartitionByMonth") \
    .getOrCreate()
```

```python
# Sample data with Unix timestamps in milliseconds
data = [
    (1, 1393670825000),  # Corresponds to 2023-09
    (2, 1353990193000),  # Corresponds to 2023-10
    (3, 1350939647000),  # Corresponds to 2023-11
]

columns = ["id", "timestamp"]
df = spark.createDataFrame(data, columns)
```

```python
df.show()
```

```python
df = df.withColumn("month_year", date_format(from_unixtime(col("timestamp") / 1000), "yyyy-MM"))
df = df.withColumn("month_year_day", date_format(from_unixtime(col("timestamp") / 1000), "yyyy-MM-dd"))
```
